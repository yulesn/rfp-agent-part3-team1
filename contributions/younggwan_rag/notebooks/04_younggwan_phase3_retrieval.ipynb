# 01. RAG 색인 구축과 원문 검색

가상 문서로 실행 경로를 확인합니다. 실제 문서 성능 측정이 아닙니다. 청킹·색인 구현은 `rfp_rag.workspace`, 검색 구현은 `rfp_rag.workflows`에 있습니다.

```python
from pathlib import Path
from rfp_rag.workspace import build_demo_workspace, open_workspace

# 먼저 team_handoff 폴더에서 uv sync --extra notebook 을 실행하고 .venv 커널을 선택합니다.
# 기본은 가상 문서입니다. 실제 자료는 CLI build --source로 별도 workspace를 만드세요.
WORKSPACE = Path.cwd() / "workspaces" / "notebook-demo"
build_demo_workspace(WORKSPACE)
```

```python
from rfp_rag.workflows import search
with open_workspace(WORKSPACE) as env:
    print(env["index"].get_catalog())
    hits = search(env, "사업예산과 안정화를 포함한 기간은?", doc_ids=["DEMO_A"])
    for hit in hits:
        print(hit["doc_id"], hit["locator"], hit["text"])
```

실제 자료: `uv run rfp-rag build --source /절대경로/source_documents.jsonl --workspace workspaces/team-v1 --config configs/rag.yaml`. 입력 계약은 README를 참고하세요.
