```python
from pathlib import Path
from rfp_rag.workspace import build_demo_workspace, open_workspace

# 먼저 team_handoff 폴더에서 uv sync --extra notebook 을 실행하고 .venv 커널을 선택합니다.
# 기본은 가상 문서입니다. 실제 자료는 CLI build --source로 별도 workspace를 만드세요.
WORKSPACE = Path.cwd() / "workspaces" / "notebook-demo"
build_demo_workspace(WORKSPACE)
```

```python
from rfp_rag.workflows import summarize
from rfp_rag.summary_full.engine import render_result
RUN_LIVE = False
RUN_ID = "summary-01"  # 모델/설정/코드를 바꾸면 새 ID와 새 workspace 사용
with open_workspace(WORKSPACE) as env:
    runner, results = summarize(env, ["DEMO_A"], run_id=RUN_ID, live=RUN_LIVE)
    for result in results:
        print(render_result(result))
    print("원문·검토표:", runner.folder)
    print("누적 실행 예산:", runner.budget_usage())
```
