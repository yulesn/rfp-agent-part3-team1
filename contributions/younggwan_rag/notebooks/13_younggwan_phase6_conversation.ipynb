```python
from pathlib import Path
from rfp_rag.workspace import build_demo_workspace, open_workspace

# 먼저 team_handoff 폴더에서 uv sync --extra notebook 을 실행하고 .venv 커널을 선택합니다.
# 기본은 가상 문서입니다. 실제 자료는 CLI build --source로 별도 workspace를 만드세요.
WORKSPACE = Path.cwd() / "workspaces" / "notebook-demo"
build_demo_workspace(WORKSPACE)
```

```python
from rfp_rag.pilot.pilot import PilotSession, session_report
from rfp_rag.workflows import submit_questions
if "env" in globals():
    env["index"].close()
env = open_workspace(WORKSPACE)
if "pilot" not in globals() or pilot.root != WORKSPACE.resolve():
    pilot = PilotSession(env)
else:
    pilot.attach(env)
```

```python
RUN_LIVE = False
BATCH_ID = "qa-01"  # 새 질문 묶음은 새 ID를 사용합니다.
DOC_IDS = ["DEMO_A"]
QUESTIONS = ["사업 예산과 부가세 조건은?", "그 기간에 안정화도 포함돼?"]
records = submit_questions(pilot, QUESTIONS, doc_ids=DOC_IDS, live=RUN_LIVE, batch_id=BATCH_ID)
for record in records:
    print(record["result"].get("markdown", ""))
    print(record["result"].get("execution_status"), record["usage_delta"])
print(session_report(pilot.folder))
```
